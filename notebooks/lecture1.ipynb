__Problem statement:__ We want to calculate the L2-norm of a vector

__Prompt:__ _Write a C++ program that computes the L2-norm of a vector_

# Result from Gemini Pro 3.1

__Gemini explanation:__ Here is a clean, modern C++ implementation using `std::valarray` / `std::vector` and `std::hypot` (or `std::inner_product`) from standard headers. Using `std::hypot` is the most numerically stable method because it avoids unwanted underflow or overflow when squaring very small or very large numbers.

````c++
#include <iostream>
#include <vector>
#include <numeric>
#include <cmath>

// Computes the L2-norm (Euclidean norm) of a vector using std::hypot
template <typename T>
double compute_l2_norm(const std::vector<T>& vec) {
    // std::hypot with std::accumulate ensures numerical stability
    return std::accumulate(vec.begin(), vec.end(), 0.0, [](double acc, T val) {
        return std::hypot(acc, val);
    });
}

int main() {
    std::vector<double> vec = {3.0, 4.0, 12.0};

    double norm = compute_l2_norm(vec);

    std::cout << "Vector: [3.0, 4.0, 12.0]\n";
    std::cout << "L2-norm: " << norm << std::endl; // Output: 13

    return 0;
}
````

# Result from Gemini Pro 3.1 - alternative

__Gemini explanation:__ If performance is a higher priority than extreme numerical stability, you can use `std::inner_product` with `std::sqrt`.

````c++
#include <iostream>
#include <vector>
#include <numeric>
#include <cmath>

template <typename T>
double compute_l2_norm_fast(const std::vector<T>& vec) {
    double sum_of_squares = std::inner_product(vec.begin(), vec.end(), vec.begin(), 0.0);
    return std::sqrt(sum_of_squares);
}

int main() {
    std::vector<double> vec = {3.0, 4.0, 12.0};

    double norm = compute_l2_norm(vec);

    std::cout << "Vector: [3.0, 4.0, 12.0]\n";
    std::cout << "L2-norm: " << norm << std::endl; // Output: 13

    return 0;
}
````

__Prompt:__ _How can I optimize the C++ L2-norm computation using SIMD or C++20 parallel algorithms?_

# Result from Gemini Pro 3.1 - performance optimized

__Gemini explanation:__ Standard C++ provides built-in parallel algorithms via standard execution policies (`std::execution::par` or `std::execution::par_unseq`). This automatically distributes the work across available CPU threads without requiring third-party libraries.

````c++
#include <iostream>
#include <vector>
#include <numeric>
#include <cmath>
#include <execution> // Required for execution policies

double l2_norm_parallel(const std::vector<double>& vec) {
    // std::transform_reduce scales across threads and auto-vectorizes
    double sum_of_squares = std::transform_reduce(
        std::execution::par_unseq, // Enables parallel execution and SIMD vectorization
        vec.begin(), vec.end(),
        0.0,
        std::plus<double>{},
        [](double x) { return x * x; }
    );

    return std::sqrt(sum_of_squares);
}

int main() {
    std::vector<double> data(100'000'000, 2.5); // 100M elements
    
    double norm = l2_norm_parallel(data);
    std::cout << "L2-Norm: " << norm << "\n";
    
    return 0;
}
````

# Intermezzo: Terminology

- A __declaration__ introduces the name of the variable or constant, and describes its type but does not create it
  ```cpp

 # auto vs. explicit types
 
- __Recommendation:__ use the keyword `auto`
  - to improve readability of the source code
  - to improve maintainability of the source code
  - to benefit from performance gains (later in this course)

- ... unless explicit conversion is required
  - `auto a = 1.5+0.3;` is the same as `(double)1.8 = 1.8`
  - `int b = 1.5+0.3;` is the same as `(int)1.8 = 1`
        
- ... unless the C++ standard does not allow so, e.g., return type of a (pure) virtual function (later in this course)

# Step B.1: Extract the sum into a function

__Prompt:__ _Define a function called `leibniz_estimate` that takes one integer parameter (the number of terms `N`) and returns a `double` --- the estimated value of $\pi$. Move the loop and the input check from `main` into this function. ````c++
#include <iostream> 
#include <stdexcept> 
  
double leibniz_estimate(int N) { 
    if (N < 1) { 
        throw std::invalid_argument( 
            "N must be at least 1 (got a non-positive value)."); 
    } 
  
    double sum = 0.0; 
    for (int k = 0; k < N; ++k) { 
        const double sign = (k % 2 == 0) ? 1.0 : -1.0; 
        sum += sign / (2.0 * k + 1.0); 
    } 
    return 4.0 * sum; 
} 
  
int main() { 
    std::cout << "Leibniz pi estimator -- v0.2" << std::endl; 
  
    const int N = 1000; 
  
    try { 
        const double estimate = leibniz_estimate(N); 
        std::cout << "Using N = " << N << " terms" << std::endl; 
        std::cout << "Estimate: " << estimate      << std::endl; 
    } 
    catch (const std::exception& e) { 
        std::cerr << "Error: " << e.what() << std::endl; 
        return 1; 
    } 
  
    return 0; 
}
````

# Block C: Arrays, initiatlisation, and pointers

In this block you will build a small statistics tool. 

Given a sample of $N$ measurements
$$
\left[ x_0, x_1, \dots, x_{N-1} \right]
$$
we want to compute the __sample mean__ $\overline{x}$ and the __sample variance__ $\sigma^2$
$$
\overline{x} = \frac{1}{N} \sum_i x_i,\qquad
\sigma^2 = \frac{1}{N-1} =sum_i (x_i - \overline{x})^2
$$

From the variance we get the __standard deviation__ $\sigma$ as its square root.

# Step C.4: Add the variance and the standard deviation

__Prompt:__ _Add a function called `variance` that takes the same `double* data` and `int n` as mean, plus a third parameter `double mean_value` (the precomputed mean). The function returns the sample variance as a `double`. In `main`, after computing the mean, call `variance(data, n, mean_value)` to get the variance, then compute the standard deviation as `std::sqrt(variance_value)` using `<cmath>`. Print all three numbers on their own labeled lines._

````c++
#include <iostream> 
#include <cmath> 
  
double mean(double* data, int n) { 
    double sum{}; 
    for (int i = 0; i < n; ++i) { 
        sum += data[i]; 
    } 
    return sum / n; 
} 
  
double variance(double* data, int n, double mean_value) { 
    double sum_sq{}; 
    for (int i = 0; i < n; ++i) { 
        double diff = data[i] - mean_value; 
        sum_sq += diff * diff; 
    } 
    return sum_sq / (n-1); 
} 
  
int main() { 
    std::cout << "Sample statistics tool -- v0.2" << std::endl; 
  
    const int n = 5; 
    double data[n] = {2.1, 2.3, 1.9, 2.2, 2.0}; 
  
    const double mean_value   = mean(data, n); 
    const double var = variance(data, n, mean_value); 
    const double sd  = std::sqrt(var); 
  
    std::cout << "Mean   = " << mean_value   << std::endl; 
    std::cout << "Var    = " << var << std::endl; 
    std::cout << "StdDev = " << sd  << std::endl; 
  
    return 0; 
}
````
