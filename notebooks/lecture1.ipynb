# Example

__Problem statement:__ We want to calculate the l2-norm of a vector

__Prompt:__ _Write a C++ program that computes the L2-norm of a vector_

# Result from Gemini Pro

```c++
#include <iostream>
#include <vector>
#include <numeric>
#include <cmath>

// Computes the L2-norm (Euclidean norm) of a vector using std::hypot
template <typename T>
double compute_l2_norm(const std::vector<T>& vec) {
    // std::hypot with std::accumulate ensures numerical stability
    return std::accumulate(vec.begin(), vec.end(), 0.0, [](double acc, T val) {
        return std::hypot(acc, val);
    });
}

int main() {
    std::vector<double> vec = {3.0, 4.0, 12.0};

    double norm = compute_l2_norm(vec);

    std::cout << "Vector: [3.0, 4.0, 12.0]\n";
    std::cout << "L2-norm: " << norm << std::endl; // Output: 13

    return 0;
}
```

Here is a clean, modern C++ implementation using `std::valarray` / `std::vector` and `std::hypot` (or `std::inner_product`) from standard headers. Using `std::hypot` is the most numerically stable method because it avoids unwanted underflow or overflow when squaring very small or very large numbers.

# Result from Gemini Pro - alternative

```c++
#include <iostream>
#include <vector>
#include <numeric>
#include <cmath>

template <typename T>
double compute_l2_norm_fast(const std::vector<T>& vec) {
    double sum_of_squares = std::inner_product(vec.begin(), vec.end(), vec.begin(), 0.0);
    return std::sqrt(sum_of_squares);
}
```

If performance is a higher priority than extreme numerical stability, you can use `std::inner_product` with `std::sqrt`.

__Prompt:__ _How can I optimize the C++ L2-norm computation using SIMD or C++20 parallel algorithms?_

# Result from Gemini Pro - performance optimized

```c++
#include <iostream>
#include <vector>
#include <numeric>
#include <cmath>
#include <execution> // Required for execution policies

double l2_norm_parallel(const std::vector<double>& vec) {
    // std::transform_reduce scales across threads and auto-vectorizes
    double sum_of_squares = std::transform_reduce(
        std::execution::par_unseq, // Enables parallel execution and SIMD vectorization
        vec.begin(), vec.end(),
        0.0,
        std::plus<double>{},
        [](double x) { return x * x; }
    );

    return std::sqrt(sum_of_squares);
}

int main() {
    std::vector<double> data(100'000'000, 2.5); // 100M elements
    
    double norm = l2_norm_parallel(data);
    std::cout << "L2-Norm: " << norm << "\n";
    
    return 0;
}
```

Standard C++ provides built-in parallel algorithms via standard execution policies (`std::execution::par` or `std::execution::par_unseq`). This automatically distributes the work across available CPU threads without requiring third-party libraries.

# Block A: Estimating $\pi$ with the Leibnitz series

The Leibnitz formula is the alternating series

$$
\frac{\pi}{4}
=
\sum_{k=0}{\infty} \frac{(-1)^k}{2k+1}
=
1-\frac{1}{3}+\frac{1}{5}-\frac{1}{7}+\frac{1}{9}-\dots
$$

 # auto vs. explicit types
 
- __Recommendation:__ use the keyword `auto`
  - to improve readability of the source code
  - to improve maintainability of the source code
  - to benefit from performance gains (later in this course)

  - `auto a = 1.5+0.3;` is the same as `(double)1.8 = 1.8`
  - `int b = 1.5+0.3;` is the same as `(int)1.8 = 1`

# Debug example: divide by zero

````c++
#include <iostream>
#include <stdexcept>

// A function that might throw an exception
int divide(int numerator, int denominator) {
    if (denominator == 0) {
        throw std::invalid_argument("Denominator cannot be zero.");
    }
    return numerator / denominator;
}

try {
    int a = 10;
    int b = 0; // Intentionally set to zero to cause an exception
    int result = divide(a, b);
    std::cout << "Result is: " << result << std::endl;
} catch (const std::invalid_argument& e) {
    // Handle the exception here
    std::cerr << "Caught an exception: " << e.what() << std::endl;
}
````

Namespaces, like `std`, allow to bundle functions even with
the same function name (and interface) into logical units

````c++
namespace tudelft {
    void hello() {
      std::cout << “Hello TU Delft\n”;
    }
}
````

````c++
namespace other {
    void hello() {
      std::cout << “Hello other\n”;
    }
}
````

Functions implemented in a namespace can be called by

- providing the namespace explicitly
  ```cpp
  tudelft::hello();
  other::hello();
  ```
  
- importing the namespace into the scope
  ```cpp
  {
    using namespace tudelft;
    hello(); 
  }

  {
    using namespace other;
    hello();
  }
  ```

Namespaces can be nested

````c++
namespace tudelft {
    void hello() { std::cout << "Hello TU Delft\n"; }
    namespace eemcs {
      void hello() { std::cout << "Hello EEMCS\n"; }
    }
}

tudelft::hello();
tudelft::eemcs::hello();
````

Leading <code>::</code> goes back to outermost unnamed namespace

````c++
namespace tudelft {
    void hello() { ::hello();
                   std::cout << "TU Delft\n"; }
    namespace eemcs {
      void hello() { ::hello();
  std::cout << "EEMCS at";
  ::tudelft::hello();
            }
    }
}

void hello() { std::cout << "Hello "; }
````
