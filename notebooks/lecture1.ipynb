# Example

__Problem statement:__ We want to calculate the L2-norm of a vector

__Prompt:__ _Write a C++ program that computes the L2-norm of a vector_

# Result from Gemini Pro 3.1

__Gemini explanation:__ Here is a clean, modern C++ implementation using `std::valarray` / `std::vector` and `std::hypot` (or `std::inner_product`) from standard headers. Using `std::hypot` is the most numerically stable method because it avoids unwanted underflow or overflow when squaring very small or very large numbers.

```c++
#include <iostream>
#include <vector>
#include <numeric>
#include <cmath>

// Computes the L2-norm (Euclidean norm) of a vector using std::hypot
template <typename T>
double compute_l2_norm(const std::vector<T>& vec) {
    // std::hypot with std::accumulate ensures numerical stability
    return std::accumulate(vec.begin(), vec.end(), 0.0, [](double acc, T val) {
        return std::hypot(acc, val);
    });
}

int main() {
    std::vector<double> vec = {3.0, 4.0, 12.0};

    double norm = compute_l2_norm(vec);

    std::cout << "Vector: [3.0, 4.0, 12.0]\n";
    std::cout << "L2-norm: " << norm << std::endl; // Output: 13

    return 0;
}
```

# Result from Gemini Pro 3.1 - alternative

__Gemini explanation:__ If performance is a higher priority than extreme numerical stability, you can use `std::inner_product` with `std::sqrt`.

```c++
#include <iostream>
#include <vector>
#include <numeric>
#include <cmath>

template <typename T>
double compute_l2_norm_fast(const std::vector<T>& vec) {
    double sum_of_squares = std::inner_product(vec.begin(), vec.end(), vec.begin(), 0.0);
    return std::sqrt(sum_of_squares);
}

int main() {
    std::vector<double> vec = {3.0, 4.0, 12.0};

    double norm = compute_l2_norm(vec);

    std::cout << "Vector: [3.0, 4.0, 12.0]\n";
    std::cout << "L2-norm: " << norm << std::endl; // Output: 13

    return 0;
}
```

__Prompt:__ _How can I optimize the C++ L2-norm computation using SIMD or C++20 parallel algorithms?_

# Result from Gemini Pro 3.1 - performance optimized

__Gemini explanation:__ Standard C++ provides built-in parallel algorithms via standard execution policies (`std::execution::par` or `std::execution::par_unseq`). This automatically distributes the work across available CPU threads without requiring third-party libraries.

```c++
#include <iostream>
#include <vector>
#include <numeric>
#include <cmath>
#include <execution> // Required for execution policies

double l2_norm_parallel(const std::vector<double>& vec) {
    // std::transform_reduce scales across threads and auto-vectorizes
    double sum_of_squares = std::transform_reduce(
        std::execution::par_unseq, // Enables parallel execution and SIMD vectorization
        vec.begin(), vec.end(),
        0.0,
        std::plus<double>{},
        [](double x) { return x * x; }
    );

    return std::sqrt(sum_of_squares);
}

int main() {
    std::vector<double> data(100'000'000, 2.5); // 100M elements
    
    double norm = l2_norm_parallel(data);
    std::cout << "L2-Norm: " << norm << "\n";
    
    return 0;
}
```

__Remark:__ With Clang++ this code requires the `-fexperimental-library` flag.

__Large language models (LLMs)__ are trained on natural language, source code, documentation, and other text.

__Code is treated much like a language:__ the model learns syntax, programming idioms, APIs, algorithms, and relationships between code and natural-language descriptions.

Given a prompt and context, __the model predicts the next token__ (word or code fragment) repeatedly to construct an answer.

__Generation is probabilistic:__ the model produces plausible code, not necessarily _correct_ code.

__Modern coding agents go beyond generation:__ they can inspect existing code, generate and modify files, compile/run programs, interpret errors, test, and iteratively improve their solution.

__Key message:__ AI-generated code should be treated as a proposed solution that must be understood, tested, and validated.

# Intermezzo: Terminology

A __declaration__ introduces the name of the variable or constant, and describes its type but does not create it
```c++
extern int var;
```

A __definition__ instantiates it (=creates it)
```c++

```

An __initialization__ initializes it (=assigns a value to it)
```c+
var = 10;
```

All three steps can be combined, e.g., `int var{10}` or split across different source/header files (later in this course)

# Scope of variables

Variables are only visible in their __scope__

```c++
int main() {
    int a = 10;  // variable a is visible here
    
    {
      int b = a; // variables a and b are visible here
    }
    
    {
      int c = b; // variables a and c are visible here, b is not(!) visible here -> error
    }            
}
```

Another example

```c++
int main() {
    int a = 10;   // variable a is visible here
    
    {
      int a = 20; // new variable a only visible in inner
                  // scope, interrupts scope of outer-scope a
      std::cout << a << std::endl; // is 20
    }
    
    std::cout << a << std::endl;   // is 10
}
```

# Quiz: What's the difference?

```c++
const int N = 1000;
double err; 

double leibniz_estimate(int N, double& error) { ... }
```

`N` is passed __by value__, i.e., function `leibniz_estimate` gets its own copy. Whatever the function does internally will not change the value in the calling scope.

`err` is passed __by reference__, i.e., function `leibniz_estimate` accesses the original memory location of the variable `err` defined in the calling scope. Changes made to variable `error` inside the function immediately affect the value of `err`. This feature can be used to "return" information from a function.

# Block C: Arrays, initiatlisation, and pointers

In this block you will build a small statistics tool. 

Given a sample of $N$ measurements
$$
\left[ x_0, x_1, \dots, x_{N-1} \right]
$$
we want to compute the __sample mean__ $\overline{x}$ and the __sample variance__ $\sigma^2$
$$
\overline{x} = \frac{1}{N} \sum_i x_i,\qquad
\sigma^2 = \frac{1}{N-1} = \sum_i (x_i - \overline{x})^2
$$

From the variance we get the __standard deviation__ $\sigma$ as its square root.
