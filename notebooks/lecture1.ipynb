# Programming languages

- A __computer programming language__ is a formal notation (set of instructions) for writing computer programs
- One distinguishes between
  - __Interpreted languages__ like Python, JavaScript, ...
  - __Compiled languages__ like C, C++, Fortran, Julia, ...
- Another distinction is between
  - __Functional programming__ that treats computation as evaluation of math functions
  - __Object-oriented programming__ that treats computation as <span style="color:red">living objects</span> that have mutable <span style="color:red">data</span> and provide <span style="color:red">methods</span> to manipulate the data

# Pros and Cons of *interpreted* languages

__Python__: Hello.py

```python
1  a = 1
2  b = 2+a
3  c = 3+b
4
5  for i in range(b, 3+b):
6    a = a+1
7
8  print(c)
```

- Processes the code line by line <span style="color:red">at run time (slow!)</span> &#128557;
  - Computes `3+b` twice (in lines 3, 5) &#128557;
  - Cannot know __in advance__ that lines 5-6 are not 'used' at all &#128557;
  - Cannot reorder code &#128557;
- Can inject/load new code instructions at run time &#128512;
- Platform independent &#128512;
- Dynamic or no typing &#128512; / &#128557;

# Pros and Cons of *compiled* languages

__C++__: Hello.cxx

```cpp
1  int a = 1;
2  int b = 2+a;
3  int c = 3+b;
4
5  for (int i=b; i<=3+b; i++)
6    a + a+1
7
8  printf("%d\n", c);
```

- Pre-processes the code line by line <span style="color:red">at compile time</span> to create (optimized) <span style="color:red">run time</span> executable
  - Evaluates lines 2-3 at compile time and replaces `2+a` and `3+b` by values &#128512;
  - Eliminates 'dead code' such as lines 1-7 &#128512;
- (Very) platform dependent &#128557;
- Compiler can check types &#128512;

# Intermezzo: Terminology

- A __declaration__ introduces the name of the variable or constant, and describes its type but does not create it
  ```cpp
  extern int f;
  ```
- A __definition__ instantiates it (=creates it)
  ```cpp
  int f;
  ```
- An __initialization__ initializes it (=assigns a value to it)
  ```cpp
  f = 10;
  ```
- All three steps can be combined, e.g., `int f{10}` or split across different source/header files (later in this course)

# Scope of variables/constants

Variables/constants are only visible in their __scope__

```cpp
int main() {
    int a = 10;  // variable a is visible here
    
    {
      int b = a; // variable a is visible here
    }
    
    {
      int c = b; // variable a is visible here, b is not(!) visible here -> error
    }            
}
```

Another example

```cpp
int main() {
    int a = 10;   // variable a is visible here
    
    {
      int a = 20; // new variable a only visible in blue
                  // scope, interrupts scope of red one
      std::cout << a << std::endl; // is 20
    }
    
    std::cout << a << std::endl;   // is 10
}
```

# Mixing and conversion of types

Getting the __type__ of a variable

````c++
#include <typeinfo>

float f = 1.7f; double d = 0.7;
std::cout << typeid(f).name() << std::endl; // float
std::cout << typeid(d).name() << std::endl; // double
````

C++ converts different types automatically

````c++
std::cout << typeid(f + d).name() << std::endl; // double
````

Check if you are happy with the result

````c++
char  x = 'a'; 
float y = 1.7;
std::cout << typeid(x + y).name() << std::endl; // float???
````

C++11 introduces the `auto` keyword, which makes handling of mixed types very easy

````c++
auto x = f + d;
std::cout << typeid(x).name() << std::endl; // double
````

You can also explicitly __cast__ one type into another

````c++
auto y = f + (float)d;
std::cout << typeid(y).name() << std::endl; // float
                
auto z = (int) (f + (float)d);
std::cout << typeid(z).name() << std::endl; // int
````

 # auto vs. explicit types
 
- __Recommendation:__ use the keyword `auto`
  - to improve readability of the source code
  - to improve maintainability of the source code
  - to benefit from performance gains (later in this course)

- ... unless explicit conversion is required
  - `auto a = 1.5+0.3;` is the same as `(double)1.8 = 1.8`
  - `int b = 1.5+0.3;` is the same as `(int)1.8 = 1`
        
- ... unless the C++ standard does not allow so, e.g., return type of a (pure) virtual function (later in this course)

# Pointer hazards

Pointers that remain uninitialized can cause hazard

````c++
int* p;
std::cout <<  p << std::endl; // prints some memory address
std::cout << *p << std::endl; // prints some random content at that address
````

````c++
int * p = nullptr;
std::cout <<  p << std::endl; // is 0x0
std::cout << *p << std::endl; // yields Segmentation fault
````

You can use `nullptr` to check if a pointer can be dereferenced without problems

````c++
std::cout << (p ? *p : nullptr) << std::endl;
````

# Debug example: divide by zero

````c++
#include <iostream>
#include <stdexcept>

// A function that might throw an exception
int divide(int numerator, int denominator) {
    if (denominator == 0) {
        throw std::invalid_argument("Denominator cannot be zero.");
    }
    return numerator / denominator;
}

try {
    int a = 10;
    int b = 0; // Intentionally set to zero to cause an exception
    int result = divide(a, b);
    std::cout << "Result is: " << result << std::endl;
} catch (const std::invalid_argument& e) {
    // Handle the exception here
    std::cerr << "Caught an exception: " << e.what() << std::endl;
}
````

# Argument passing – by reference

Arguments that are passed __by reference__ give the function read and write access to the memory location of the original variable

````c++
int addOneByReference(int& a) { return a + 1; }

int i = 1;
int j = addOneByReference(i);
````

*Quiz*: What happens if `addOneByReference()` tries to modify the value of the passed variable?

The __post-increment operator__ `a++` performs the operation first and incremenets variable `a` afterwards

````c++
int addOneByReference1(int& a) { return a++; }

int i = 1;
int j = addOneByReference1(i);
````

The __pre-increment operator__ `++a` incremenets variable `a` first and performs the operation afterwards

````c++
int addOneByReference2(int& a) { return ++a; }

int i = 1;
int j = addOneByReference2(i);
````

# C++ return value optimization (RVO)
    
Most C++ compilers support RVO, that is, no temporary variable for the return value is created inside the function

```cpp
int addOne(const int& a) { return a+1; }
```

but the return value is immediately assigned to variable `j`

```cpp
int i = 1;
int j = addOne(i); // RVO makes it int j = (i+1);
```

# Argument passing

If we want a function that changes the argument __directly__, we must pass the argument by reference.

```cpp
void addOne_Val(int  a) { a++; } // increment local copy
void addOne_Ref(int& a) { a++; } // increment a(~i)
```

```cpp
int i = 1;       // i=1
addOne_Val(i);   // i=1 (still)
addOne_Ref(i);   // i=2
```

The return type `void` indicates that 'nothing' is returned.

# Example

Compute the sum of the entries of an array

````c++
double sum(const int* array, int length) {
    
    double s = 0;

    for (auto i=0; i<length; i++)
        s += array[i];

    return s;
}

int array[5] = { 1, 2, 3, 4, 5 };
std::cout << sum(array, 5) << std::endl;
````

This is not OOP. <b>DON'T, REALLY DON'T DO THIS IN C++!</b> We will learn much better ways to pass bigger objects such as arrays by (constant) reference.

There is one exception. If you want to allocate memory dynamically inside the function and assign it to a variable defined outside the function you need to work with double pointers.

# Static arrays

Definition and creation of a static array

```cpp
int array[5];
```

Definition, creation and initialization of a static array

```cpp
int array[5] = { 1, 2, 3, 4, 5 }; // since C++11
int array[5]{ 1, 2, 3, 4, 5 }; // since C++11</pre>
```

Access of individual array positions


````c++
for (auto i=0; i<5; i++)
    std::cout << array[i] << std::endl;
````

 <p>Remember that C++ <strong>starts indexing at 0</strong></p>
 <p>Static arrays are <strong>destroyed automatically</strong> at the end of scope</p>

# Dynamic arrays

- Definition and allocation of dynamic array

  ```cpp
   int* array = new int[5];
  ```

- Definition, allocation and initialization of dynamic array

  ```cpp
  int* array = new int[5]{ 1, 2, 3, 4, 5 }; // in C++11
  ```

- Explicit deallocation of dynamically allocated array needed

  ```cpp
  delete[] array;
  ```

- Think fail-safe! Because it still points to an invalid address
  ```cpp
  array = nullptr;
  ```

# Example of a dynamic array

````c++
int* array = new int[5]{ 1, 2, 3, 4, 5 };

for (auto i=0; i<5; i++)
    std::cout << array[i] << std::endl;

delete[] array;
array = nullptr;
````

# Static vs. dynamic arrays

<b>Static arrays</b> require the size to be known at compile time 

````c++
int array[5];
constexpr int k=5; // constexpr tells the compiler that the 
int array[k];      // expression is available at compile time
int k=5;
int array[k];      // Gives a compiler error !!!
````

<b>Dynamic arrays</b> allow variable sizes at run-time

````c++
int k = std::atoi(argv[1]);
int* array = new int[k];
````

# Namespaces

Namespaces, like `std`, allow to bundle functions even with
the same function name (and interface) into logical units

````c++
namespace tudelft {
    void hello() {
      std::cout << “Hello TU Delft\n”;
    }
}
````

````c++
namespace other {
    void hello() {
      std::cout << “Hello other\n”;
    }
}
````

Functions implemented in a namespace can be called by

- providing the namespace explicitly
  ```cpp
  tudelft::hello();
  other::hello();
  ```
  
- importing the namespace into the scope
  ```cpp
  {
    using namespace tudelft;
    hello(); 
  }

  {
    using namespace other;
    hello();
  }
  ```

Namespaces can be nested

````c++
namespace tudelft {
    void hello() { std::cout << "Hello TU Delft\n"; }
    namespace eemcs {
      void hello() { std::cout << "Hello EEMCS\n"; }
    }
}

tudelft::hello();
tudelft::eemcs::hello();
````

Leading <code>::</code> goes back to outermost unnamed namespace

````c++
namespace tudelft {
    void hello() { ::hello();
                   std::cout << "TU Delft\n"; }
    namespace eemcs {
      void hello() { ::hello();
  std::cout << "EEMCS at";
  ::tudelft::hello();
            }
    }
}

void hello() { std::cout << "Hello "; }
````
